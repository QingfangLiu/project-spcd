```python
import pandas as pd
from pathlib import Path

# Define input and output base directories
input_base = Path('/Users/liuq13/project-SPCD/beh_raw')
output_base = Path('/Users/liuq13/project-SPCD/beh_processed_subs')

# Track subjects with missing files
missing_subjects = []
```

```python
# Loop through each subject
for subject_dir in input_base.iterdir():
    if subject_dir.is_dir() and subject_dir.name.startswith('sub-'):
        subject_id = subject_dir.name
        ratings_dir = subject_dir / 'ratings'
        pre_file = ratings_dir / 'PrePleasantness.csv'
        post_file = ratings_dir / 'PostPleasantness.csv'

        if not pre_file.exists() or not post_file.exists():
            missing_subjects.append(subject_id)
            continue

        # Load data
        pre_df = pd.read_csv(pre_file, header=None)
        post_df = pd.read_csv(post_file, header=None)

        # Compute mean
        pre_mean = pre_df.mean(axis=1)
        post_mean = post_df.mean(axis=1)

        # Create summary DataFrame
        summary_df = pd.DataFrame({
            'odor_num': pre_df.index,
            'pre_mean': pre_mean,
            'post_mean': post_mean,
            'change': post_mean - pre_mean
        })

        # Save output
        out_dir = output_base / subject_id
        out_dir.mkdir(parents=True, exist_ok=True)
        summary_df.to_csv(out_dir / 'pleasantness_summary.csv', index=False)

        print(f"Saved summary for {subject_id}")

# Report missing subjects
if missing_subjects:
    print("\nSubjects with missing pre or post pleasantness ratings:")
    for s in missing_subjects:
        print(f" - {s}")
else:
    print("\nAll subjects processed successfully.")
```

```python
# path to the specific file
csv_path = output_base / 'sub-14' / 'pleasantness_summary.csv'

# load
df = pd.read_csv(csv_path)

# swap 1 ↔ 2 in odor_num
df['odor_num'] = df['odor_num'].replace({1: 2, 2: 1})

# save
df.to_csv(csv_path, index=False)
```
