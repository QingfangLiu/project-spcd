### Subject Info Loading, Cleaning, and Demographic Summary

This code loads the subject_info.csv file for the SPCD project, removes excluded subjects (sub-19, sub-33), and computes key demographic and timing statistics for the final analysis sample. It parses Day 0/1/2 dates, calculates inter-session intervals, and summarizes gender, feeding group, pairing group, and age distributions. The resulting dataset reflects the 32 participants retained for all downstream analyses.

```python
from pathlib import Path
import pandas as pd
```

```python
# -----------------------
# Paths & basic settings
# -----------------------
study_path = Path('/Users/liuq13/SPCD_data/task_info')
subject_csv = study_path / 'subject_info.csv'
EXCLUDE_SUBJECTS = {'sub-19', 'sub-33'}
```

```python
# -----------------------
# Load & clean
# -----------------------
df = pd.read_csv(subject_csv)

# make sure subject IDs are clean strings
df['subject'] = df['subject'].astype(str).str.strip()

# drop excluded subjects
df = df[~df['subject'].isin(EXCLUDE_SUBJECTS)].reset_index(drop=True)
```

```python
df.head()
```

```python
# -----------------------
# Parse dates & intervals
# -----------------------
date_cols = ['Day 0', 'Day 1', 'Day 2']

for col in date_cols:
    df[col] = pd.to_datetime(df[col])

df['days_d0_d1'] = (df['Day 1'] - df['Day 0']).dt.days
df['days_d1_d2'] = (df['Day 2'] - df['Day 1']).dt.days
df['days_d0_d2'] = (df['Day 2'] - df['Day 0']).dt.days
```

```python
df.head()
```

```python
# -----------------------
# Basic summaries
# -----------------------
print(f'N subjects (after exclusion): {len(df)}\n')

print('Gender counts:')
print(df['gender'].value_counts().sort_index(), '\n')

print('Feeding (meal) group counts:')
print(df['feeding'].value_counts().sort_index(), '\n')

print('Pairing group counts:')
print(df['pairing'].value_counts().sort_index(), '\n')

print('Feeding × Pairing contingency table:')
print(pd.crosstab(df['feeding'], df['pairing']), '\n')

print('Age summary:')
print(df['age'].describe(), '\n')

print('Day-interval summaries (in days):')
print(df[['days_d0_d1', 'days_d1_d2', 'days_d0_d2']].describe(), '\n')

# optional: save a version with derived columns
# df.to_csv(study_path / 'subject_info_with_derived.csv', index=False)
```
