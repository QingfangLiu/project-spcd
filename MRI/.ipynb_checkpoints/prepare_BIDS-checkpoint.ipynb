```python
import os
import shutil
import re
from pathlib import Path
```

```python
# Path to the directory containing Sub*_done folders
base_dir = Path("/Users/liuq13/SPCD_data")   
BIDS_dir = Path("/Users/liuq13/SPCD_data_BIDS")  

# List all subject folders
subject_folders = sorted([f for f in base_dir.glob("Sub*_done") if f.is_dir()])
print(subject_folders)
```

```python
len(subject_folders)
```

```python
# Define task names
task_list = ["odorpred", "place", "postmeal", "premeal", "character", "odor_scan"]  # adjust if needed
```

```python
for subj_path in subject_folders:
    subj_id = re.search(r"Sub(\d+)", subj_path.name).group(1)
    bids_subj = f"sub-{int(subj_id):02d}"
    
    # Create BIDS subject directory
    func_bids_dir = BIDS_dir / bids_subj / "func"
    func_bids_dir.mkdir(parents=True, exist_ok=True)

    func_raw = subj_path / "nii" / "func_raw"
    
    for task in task_list:
        task_dir = func_raw / task
        if not task_dir.exists():
            continue
        
        for f in sorted(task_dir.glob("*.nii*")):
            # Identify echo number from filename
            echo_match = re.search(r"echo-(\d+)", f.name)
            if not echo_match:
                # Try fallback pattern
                echo_match = re.search(r"echo(\d+)", f.name)
            if echo_match:
                echo = echo_match.group(1)
            else:
                print(f"Skipping (no echo number found): {f}")
                continue

            # Determine run number (assume run-1 unless more info)
            run = "1"

            new_name = f"{bids_subj}_task-{task}_run-{run}_echo-{echo}_bold.nii.gz"
            new_path = func_bids_dir / new_name

            shutil.copy(f, new_path)
```
