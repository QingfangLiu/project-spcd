```python
import os

# Set path to sub-07's func_PAID folder
base_path = '/Users/liuq13/SPCD_data/nii/sub-07/func_PAID'  # Adjust path as needed

# Loop through all task folders (e.g., character, odor_scan, odorpred)
for task_name in os.listdir(base_path):
    task_path = os.path.join(base_path, task_name)
    if not os.path.isdir(task_path):
        continue

    for fname in os.listdir(task_path):
        if fname.startswith('s6'):
            file_path = os.path.join(task_path, fname)
            try:
                os.remove(file_path)
                print(f"Deleted: {file_path}")
            except Exception as e:
                print(f"Failed to delete {file_path}: {e}")
```
