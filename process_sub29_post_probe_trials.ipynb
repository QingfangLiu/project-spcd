```python
import pandas as pd
from pathlib import Path

# Define paths
base_dir = Path('/Users/liuq13/SPCD_data/beh_raw/sub-29/before_cleaning')
subject_dir = Path('/Users/liuq13/SPCD_data/beh_raw/sub-29')

# Input files
earlier_file = base_dir / 'Sub29_Day2_Probe_post_2025-03-08_15h05.53.365ProbeTrials.csv'
later_file = base_dir / 'Sub29_Day2_Probe_post_2025-03-08_15h41.43.522ProbeTrials.csv'

# Load data
df_early = pd.read_csv(earlier_file)
df_late = pd.read_csv(later_file)

# Coerce Run column to numeric
df_early['Run'] = pd.to_numeric(df_early['Run'], errors='coerce')
df_late['Run'] = pd.to_numeric(df_late['Run'], errors='coerce')

# Extract run 1 and 2 from early file
run1_and_2 = df_early[df_early['Run'].isin([1, 2])]

# Extract run 1 from later file and recode to Run = 3
run3 = df_late[df_late['Run'] == 1].copy()
run3['Run'] = 3

# Combine
combined_df = pd.concat([run1_and_2, run3], ignore_index=True)

# Save using the name of the later file, but to the subject folder
output_file = subject_dir / later_file.name
combined_df.to_csv(output_file, index=False)

print(f"✅ Combined and recoded file saved to:\n{output_file}")
```
