```python
import os
import glob
import subprocess
```

```python
# Set home directory
HomeDir = '/Users/liuq13/SPCD_data'

# Get list of folders that start with 'Sub' in the directory
SubIDlist = [f for f in glob.glob(os.path.join(HomeDir, 'Sub*_done')) if os.path.isdir(f)]

# Get the number of subjects
nSubIDlist = len(SubIDlist)
```

```python
SubIDlist
```

```python
for subj in range(nSubIDlist):
    SubPath = SubIDlist[subj]
    SubID = os.path.basename(SubPath)
    print(f'Processing: {SubID}')
    SubDir = os.path.join(HomeDir, SubID)
    anat_dir = os.path.join(SubDir, 'nii', 'anat')

    # Get anatomical image in the native space
    search_pattern = os.path.join(anat_dir, 'Scan*MPRAGE*.nii')
    anat_files = [f for f in glob.glob(search_pattern) if 'defaced' not in f.lower()]
    
    if anat_files:
        afile = anat_files[0]  # Use the first match
        print(f'  Found anatomical file: {afile}')

        # Run pydeface on the anatomical file
        # if done already, will return error
        try:
            subprocess.run(['python3', '-m', 'pydeface', afile], check=True)
            print(f'  Defaced successfully: {afile}')
        except subprocess.CalledProcessError as e:
            print(f'  Error during defacing: {afile}\n    {e}')
    else:
        print(f'  No anatomical file found for {SubID}')
```
