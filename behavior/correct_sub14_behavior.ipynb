### Correction Summary for sub-14 Behavioral Data

Due to a meal assignment error for participant `sub-14`, the following corrections were applied  
to their behavioral data files (`pre_meal_dat.csv` and `post_meal_dat.csv`):

1. **Swapped `choice_class` Labels:**
   - `"B1>B3"` ↔ `"B2>B3"`
   - `"A1>A3"` ↔ `"A2>A3"`

2. **Re-coded Accuracy Variables for Specific Conditions:**
   - For trials where `choice_class` is `"B1>B2"` or `"A1>A2"`:
     - `correct`: Swapped 0 and 2  
     - `acc`: Swapped 0 and 1 **only if** `resp` was present (not missing)

3. **Output:**
   - Corrected files replaced the original versions  
   - Original files were manually moved to the `before_correction` folder  
   - Location: `/Users/liuq13/SPCD_data/beh_processed/sub-14/`


```python
import pandas as pd
import numpy as np
from pathlib import Path

# Set paths
base_dir = Path("/Users/liuq13/SPCD_data/beh_processed/sub-14")
files = ["pre_meal_dat.csv", "post_meal_dat.csv"]

# Choice class swaps
swap_classes = {
    "B1>B3": "B2>B3",
    "B2>B3": "B1>B3",
    "A1>A3": "A2>A3",
    "A2>A3": "A1>A3"
}
```

```python
# Function to apply corrections
def correct_behavioral_data(df):
    # Swap choice_class
    df["choice_class"] = df["choice_class"].replace(swap_classes)

    # For selected choice_classes, modify 'correct' and 'acc'
    mask = df["choice_class"].isin(["B1>B2", "A1>A2"])
    
    # Swap correct 0 <-> 2
    df.loc[mask & (df["correct"] == 0), "correct"] = 99  # temp value
    df.loc[mask & (df["correct"] == 2), "correct"] = 0
    df.loc[mask & (df["correct"] == 99), "correct"] = 2

    # Swap acc 0 <-> 1, but only when resp is not NaN
    acc_mask = mask & df["resp"].notna()
    df.loc[acc_mask & (df["acc"] == 0), "acc"] = 99
    df.loc[acc_mask & (df["acc"] == 1), "acc"] = 0
    df.loc[acc_mask & (df["acc"] == 99), "acc"] = 1

    return df
```

```python
# Process each file
for fname in files:
    df = pd.read_csv(base_dir / fname)
    df_corrected = correct_behavioral_data(df)
    corrected_name = fname.replace(".csv", "_corrected.csv")
    df_corrected.to_csv(base_dir / corrected_name, index=False)

print("Correction completed and new files saved.")
```

```python
file_path = base_dir / "odor_pred_dat.csv"

# Load the file
df = pd.read_csv(file_path)

# Swap 1 and 2 in odor_num
df['odor_num'] = df['odor_num'].replace({1: -1, 2: 1})  # temp replace to avoid overwrite conflict
df['odor_num'] = df['odor_num'].replace({-1: 2})

# Save corrected file
corrected_path = base_dir / "odor_pred_dat_corrected.csv"
df.to_csv(corrected_path, index=False)

print(f"✅ Saved corrected file to: {corrected_path}")
```
